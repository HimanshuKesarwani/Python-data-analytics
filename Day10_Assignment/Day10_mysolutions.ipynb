```python
import pandas as pd
df1 = pd.read_csv('orders.txt')
df1
```

```python
df2 = pd.read_csv('returns.txt')
df2
```

```python
#1- write a program to get category wise sales of orders that were not returned
dfr=pd.merge(left = df1 , right = df2 ,how = 'left', left_on = 'order_id' , right_on = 'order_id')
fc = dfr['Return_Reason'].isna() #gives a boolean array for all the not returned orders
df1[fc].groupby('category', as_index=False)['sales'].sum()
#first did left join on 'orders' and 'returns' table on 'order_id' and then check for isnull on 'Return Reason' column and applied the filter condition 
#after applying filter condition group by category and on sum of sales
```

```python
#2- write a program to get city wise count of return orders
dfr=pd.merge(left = df1 , right = df2 ,how = 'inner', left_on = 'order_id' , right_on = 'order_id')
dfr['city'].value_counts()
#above script gives citywise count of return orders and changes the index column from 'Return_Reason' to 'return_order_count'
```

```python
#3- write a program to print cities where we have all 3 kinds of returns (others,bad quality,wrong items)
dfr=pd.merge(left = df1 , right = df2 ,how = 'inner', left_on = 'order_id' , right_on = 'order_id')
city_return_ucounts = dfr.groupby('city')['Return_Reason'].nunique() # Counts the uniques values of Return_Reason for each city 
city_return_ucounts[city_return_ucounts.values > 1]
```

```python
#4- write a program to find cities where not even a single order was returned.
dfr=pd.merge(left = df1 , right = df2 ,how = 'left', left_on = 'order_id' , right_on = 'order_id')
fc = dfr['Return_Reason'].isna() #gives a boolean array for all the not returned orders
df1[fc]['city']
```

```python
#5- write a program to find top 3 cities by sales
df1.groupby('city', as_index=False)['sales'].sum().sort_values(by = 'sales', ascending = False).head(3)
```

```python
#6- write a program to get order ids whos return reason is not known (nan)
fc = df2['Return_Reason'].isnull()
df2[fc]
```

```python
#7- write a program to fill unknown return reason to a default value 'others'
df2['Return_Reason'] = df2['Return_Reason'].fillna(value='others')
df2
```

```python
#8- write a program to find avg sales and avg profit for each category
df1.groupby('category', as_index=False).agg({'sales':'mean','profit':'mean'})
```
